1. Write a program to enter Name and Percentage Marks in a dictionary and display the information.

```python
student = {}
n = int(input()) # no of students

for i in range(n):
    name = input()
    marks = int(input())

    student[name] = marks
print('\n')

for name, percentage in student.items():
    print(name,percentage)
```

2. Write a program to find the number of occurrences of each letter in a given string.

```python
s = input()
freq = {}
for ch in s:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

for c, cnt in freq.items():
    print(c,cnt)
```

3. Write a program to find the number of occurrences of each vowel in a given string.

```python
st = input().lower()
vowels = {
    'a' : 0,
    'e' : 0,
    'i' : 0,
    'o' : 0,
    'u' : 0
}

for ch in st:

    if ch in vowels:
        vowels[ch] += 1


for c, cnt in vowels.items():
    print(c, cnt)
```

4. Write a program to accept student names and marks from the keyboard and create a dictionary. Also display the student marks by taking the student name as input.

```python
n = int(input())
students = {}

for i in range(n):
    name = input()
    marks = int(input())

    students[name] = marks

print("Enter Search Query")
search_query = input()

if search_query in students:
    print(f"Marks: {students[search_query]}")
else:
    print('Invalid query')
```
